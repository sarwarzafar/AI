Multiple Linear Regression

Importing Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

Importing Dataset

```python
dataset = pd.read_csv('50_Startups.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

```python
print(X)
```

```python
print(y)
```

Encoding Categorical Data

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [3])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
```

```python
print(X)
```

Splitting Dataset into Train and Test Set

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
```

```python
print(X_train)
```

```python
print(X_test)
```

```python
print(y_train)
```

```python
print(y_test)
```

Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

X_train[:, 3:] = sc.fit_transform(X_train[:, 3:])
X_test[:, 3:] = sc.fit_transform(X_test[:, 3:])
```

```python
print(X_train)
```

```python
print(X_test)
```

Multiple Linear Regression

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

Predicting Test Results

```python
# Predict
y_pred = regressor.predict(X_test)
# Display Precision
np.set_printoptions(precision=2)
# Convert 1D Arrays into columns
predicted = y_pred.reshape(len(y_pred), 1)
actual = y_test.reshape(len(y_test), 1)
# Combine them side by side
results = np.concatenate((predicted, actual), axis = 1)
# print comparison
print(results)
```

```python
startup = [[160000, 130000, 300000, 'California']]
startup_encoded = ct.transform(startup)
prediction = regressor.predict(startup_encoded)
print("predicted profit: ", prediction[0])
```
